```python
import pandas as pd

df = pd.read_csv("olist_closed_deals_dataset.csv")
df.head()
```

```python
print(df.columns)
```

```python
print(df.columns)
```

```python
import numpy as np


df['Campaign_ID'] = range(1, len(df)+1)


df['Customer_Segment'] = np.random.choice(['Small Business', 'Medium Business', 'Enterprise'], size=len(df))


df['Campaign_Cost'] = np.random.randint(1000, 10000, len(df))


df['Impressions'] = np.random.randint(5000, 50000, len(df))

df['Clicks'] = (df['Impressions'] * np.random.uniform(0.05, 0.2, len(df))).astype(int)


df['Conversions'] = (df['Clicks'] * np.random.uniform(0.05, 0.3, len(df))).astype(int)


df['Revenue'] = df['Conversions'] * np.random.randint(100, 500, len(df))
```

```python
df[['Campaign_ID','Customer_Segment','Campaign_Cost','Impressions','Clicks','Conversions','Revenue']].head()
```

```python

df['CTR'] = df['Clicks'] / df['Impressions']


df['Conversion_Rate'] = df['Conversions'] / df['Clicks']


df['CPA'] = df['Campaign_Cost'] / df['Conversions']


df['ROAS'] = df['Revenue'] / df['Campaign_Cost']
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
sns.barplot(x='Campaign_ID', y='ROAS', data=df)
plt.title("ROAS by Campaign")
plt.show()
```

```python
sns.scatterplot(x='CTR', y='Revenue', data=df)
plt.title("CTR vs Revenue")
plt.show()
```

```python
sns.barplot(x='Customer_Segment', y='Revenue', data=df)
plt.title("Revenue by Customer Segment")
plt.show()
```

```python
sns.scatterplot(x='Campaign_Cost', y='Conversions', data=df)
plt.title("Cost vs Conversions")
plt.show()
```

```python

bad_campaigns = df[(df['Campaign_Cost'] > df['Campaign_Cost'].mean()) &
                   (df['Revenue'] < df['Revenue'].mean())]

print(bad_campaigns[['Campaign_ID','Campaign_Cost','Revenue']])


print(df.groupby('Customer_Segment')['ROAS'].mean())


print(df[['CTR','Revenue']].corr())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, r2_score

X = df[['Campaign_Cost', 'Impressions', 'Clicks']]
y = df['Conversions']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = DecisionTreeRegressor()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```
